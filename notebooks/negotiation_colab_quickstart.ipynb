### 4-1. 실험 당시 소스 준비

아래 셀은 당시 소스 ZIP이 있는 경우에만 실행할 수 있습니다.

```
#@title 1. 소스 ZIP 선택 · 실행 폴더 준비
import hashlib, json, os, stat, subprocess, sys, uuid, zipfile
from datetime import datetime, timezone
from pathlib import Path, PurePosixPath

NEG_SOURCE_COMMIT = '2f57d074c039646dde96ad1afa9e90247b87037a'
NEG_CONTENT_HASH = '8d75ff31491e21bda062b77d39b34d542a075fbc280501c2da5745ae251f269c'
NEG_READY = False
NEG_RESULTS = {}
if not Path('/content').is_dir():
    raise RuntimeError('이 셀은 Google Colab에서 실행하세요.')
from google.colab import files
neg_archive = Path('/content/negotiation-source-2f57d074c039.zip')
if neg_archive.is_file():
    print('파일 패널에 업로드된 소스 ZIP을 확인합니다.')
else:
    neg_uploaded = files.upload()
    if len(neg_uploaded) != 1:
        raise ValueError('협상 소스 ZIP 한 개를 선택하세요. 파일 선택창이 열리지 않으면 왼쪽 파일 패널로 업로드한 뒤 다시 실행하세요.')
    neg_archive = Path(next(iter(neg_uploaded)))
    del neg_uploaded
NEG_REPO = Path('/content') / ('negotiation-' + uuid.uuid4().hex[:12])
with zipfile.ZipFile(neg_archive) as bundle:
    members = [m for m in bundle.infolist() if not m.is_dir()]
    roots = [m.filename[:-len('pyproject.toml')] for m in members
             if m.filename == 'pyproject.toml' or m.filename.endswith('/pyproject.toml')]
    if len(roots) != 1 or sum(m.file_size for m in members) > 32 * 1024**2:
        raise ValueError('예상한 프로젝트 ZIP이 아닙니다. 위 링크의 파일을 사용하세요.')
    prefix = roots[0]
    contents = {}
    for member in members:
        if not member.filename.startswith(prefix):
            raise ValueError('ZIP의 프로젝트 경로를 확인할 수 없습니다.')
        name = member.filename[len(prefix):]
        path = PurePosixPath(name)
        if (not name or '\\' in name or ':' in name or path.is_absolute()
                or '..' in path.parts or name in contents
                or stat.S_ISLNK(member.external_attr >> 16)):
            raise ValueError('ZIP에 허용하지 않는 경로가 있습니다.')
        contents[name] = bundle.read(member)
    content_hash = hashlib.sha256(''.join(
        f'{n}\0{hashlib.sha256(contents[n]).hexdigest()}\n' for n in sorted(contents)
    ).encode()).hexdigest()
    if content_hash != NEG_CONTENT_HASH:
        raise ValueError('실험 당시 소스와 다릅니다. 4-1의 고정 ZIP을 선택하세요.')
    NEG_REPO.mkdir()
    for name, data in contents.items():
        target = NEG_REPO / name
        target.parent.mkdir(parents=True, exist_ok=True)
        target.write_bytes(data)
del contents
NEG_RUN = NEG_REPO / 'runs' / 'negotiation' / ('colab-' + uuid.uuid4().hex[:8])
NEG_RUN.mkdir(parents=True)
NEG_ENV = os.environ.copy()
NEG_ENV['PYTHONPATH'] = os.pathsep.join([str(NEG_REPO / 'src'), str(NEG_REPO / 'negotiation/src')])
NEG_ENV['PYTHONUNBUFFERED'] = '1'
NEG_ENV['PYTHONIOENCODING'] = 'utf-8'
(NEG_RUN / 'source.json').write_text(json.dumps({
    'source_commit': NEG_SOURCE_COMMIT, 'content_sha256': content_hash,
    'zip_sha256': hashlib.sha256(neg_archive.read_bytes()).hexdigest(),
    'created_at': datetime.now(timezone.utc).isoformat(),
    'notice': 'New development run; separate from the saved 2026-11-16 measurements.'
}, indent=2), encoding='utf-8')

def neg_run(*args):
    command = [sys.executable, '-u', *map(str, args)]
    log = NEG_RUN / ('command-' + uuid.uuid4().hex[:8] + '.log')
    print('실행:', ' '.join(command), flush=True)
    # The environment may contain HF_TOKEN; never serialize or print it.
    env = {**os.environ, **{k: NEG_ENV[k] for k in ('PYTHONPATH', 'PYTHONUNBUFFERED', 'PYTHONIOENCODING')}}
    with log.open('w', encoding='utf-8') as output:
        process = subprocess.Popen(command, cwd=NEG_REPO, env=env,
                                   stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                                   text=True, encoding='utf-8', errors='replace')
        try:
            for line in process.stdout:
                output.write(line); output.flush()
                print(line, end='', flush=True)
            exit_code = process.wait()
        except KeyboardInterrupt:
            process.terminate()
            try:
                process.wait(timeout=15)
            except subprocess.TimeoutExpired:
                process.kill(); process.wait()
            raise
    with (NEG_RUN / 'commands.jsonl').open('a', encoding='utf-8') as history:
        history.write(json.dumps({'argv': command, 'exit_code': exit_code, 'log': log.name}) + '\n')
    if exit_code:
        raise RuntimeError(f'실행 실패(exit={exit_code}). 위 오류와 {log.name}을 확인하세요.')

def neg_experiment(name, script, *options):
    if not NEG_READY:
        raise RuntimeError('4-2~4-4의 모델 준비를 먼저 완료하세요.')
    NEG_RESULTS.pop(name, None)
    attempt = NEG_RUN / (name + '-' + uuid.uuid4().hex[:8])
    attempt.mkdir()
    result = attempt / 'result.json'
    neg_run(script, *options, *NEG_ARGS, '--output', result)
    data = json.loads(result.read_text(encoding='utf-8'))
    if not data.get('complete'):
        raise RuntimeError('실험이 완료되지 않았습니다. 결과 파일과 오류 기록을 확인하세요.')
    NEG_RESULTS[name] = result
    print('완료:', name, '| 결과:', result)
    return data

print('소스 확인 완료:', NEG_SOURCE_COMMIT[:7])
print('결과 폴더:', NEG_RUN)
```

### 4-8. 결과 표 확인 · 파일 내려받기

표에는 **이번에 완료한 실험**만 표시됩니다. 후보 실험 또는 개발 30건 중 하나만 실행해도 해당 결과를 볼 수 있습니다. 2번의 기존 측정값과 비교하되, 모델 실행 환경·반복 횟수·90% 구간을 함께 확인하세요.

그다음 셀은 이번 실행의 JSON·로그만 ZIP으로 내려받습니다. 모델 가중치·로그인 캐시는 포함하지 않습니다. 오류가 났을 때도 로그를 보관할 수 있습니다. 런타임을 종료하기 전에 내려받으세요.

```
#@title 8. 이번 실행의 결과를 표로 보기 (모델 호출 없음)
from IPython.display import Markdown, display

def neg_show_results(paths):
    rows = []
    if 'probe' in paths:
        p = json.loads(Path(paths['probe']).read_text(encoding='utf-8'))
        m = p['metrics']
        rows += [('후보 실험 완료', str(p['complete'])),
                 ('후보 점수 판정', 'C/D · 후보 점수만으로 채택할 근거 부족' if p['verdict']['route'].startswith('C_or_D') else p['verdict']['route']),
                 ('후보 순서 비교 정확도 · 모델 / 규칙',
                  f"{m['discrimination']['within_state_pair_accuracy']:.1%} / {m['reference_rule_discrimination']['within_state_pair_accuracy']:.1%}"),
                 ('후보 점수로 고를 때 규칙 대비 평균 차이', f"{m['decision_value']['gain_vs_rule_mean']:+,.0f}원")]
    if 'dev30' in paths:
        d = json.loads(Path(paths['dev30']).read_text(encoding='utf-8'))
        a, r = d['policies']['agent'], d['policies']['rule_policy']
        delta = d['agent_minus']['rule_policy']['realized_saving']
        lo, hi = delta['ci90']
        rows += [('개발 협상 완료 / 사례 수', f"{d['complete']} / {a['cases']}건"),
                 ('평균 실현 절감 · 모델 / 규칙', f"{a['mean_realized_saving']:,.0f}원 / {r['mean_realized_saving']:,.0f}원"),
                 ('규칙 대비 평균 차이 · 90% 구간', f"{delta['mean_difference']:+,.0f}원 · [{lo:+,.0f}, {hi:+,.0f}]원"),
                 ('권한 위반 / 대안보다 불리한 합의', f"{a['authority_violations']} / {a['agreements_not_better_than_alternative']}"),
                 ('사용한 근거 유형', d['evidence_class'])]
        display(Markdown('**차이 해석:** ' + ('90% 구간에 0이 포함되어 규칙보다 낫다고 확정할 수 없습니다.'
                          if lo <= 0 <= hi else '이 개발 표본의 차이이며, 새 사례에서의 성능을 보장하지 않습니다.')))
    if not rows:
        print('아직 전체 실험 결과가 없습니다. 4-6 또는 4-7을 실행한 뒤 다시 누르세요.')
        return
    display(Markdown('| 이번 실행에서 확인한 것 | 결과 |\n|---|---|\n' +
                     '\n'.join(f'| {key} | {value} |' for key, value in rows)))
    print('합성 개발 실험 결과입니다. 저장된 2026-11-16 결과와 별도로 해석하세요.')

neg_show_results(NEG_RESULTS)
```
